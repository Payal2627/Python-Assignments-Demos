```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/cars93.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
avg_price = df.groupby('Manufacturer')['Price'].mean()
plt.figure(figsize=(12,6))
plt.bar(avg_price.index,
        avg_price.values,
        color=plt.cm.tab20.colors)
plt.xlabel("Manufacturer")
plt.ylabel("Average Price")
plt.title("Average Price of Cars by Manufacturer")
plt.xticks(rotation=90)
plt.show()
```

```python
hp_type = df.groupby('Type')['Horsepower'].mean()
plt.figure(figsize=(8,5))
plt.plot(hp_type.index,
         hp_type.values,
         marker='o',
         linewidth=2)
plt.xlabel("Car Type")
plt.ylabel("Average Horsepower")
plt.title("Horsepower Trend by Car Type")
plt.legend(["Horsepower"])
plt.grid(True)
plt.show()
```

```python
type_count = df['Type'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(type_count,
        labels=type_count.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Distribution of Car Types")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Horsepower'],
            df['MPG.city'])
plt.xlabel("Horsepower")
plt.ylabel("MPG City")
plt.title("Horsepower vs MPG")
plt.show()
```

```python
hp_type = df.groupby('Type')['Horsepower'].mean()
plt.figure(figsize=(8,5))
plt.plot(hp_type.index,
         hp_type.values,
         marker='o')
plt.xlabel("Car Type")
plt.ylabel("Horsepower")
plt.title("Trend of Horsepower by Car Type")
plt.legend(["Horsepower"])
plt.show()
```
