```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/tips.csv")
print(df.head())
```

```python
# 1. Day-wise Boxplot of Total Bill
plt.figure(figsize=(8,5))
sns.boxplot(x='day', y='total_bill', data=df)
plt.title("Day-wise Boxplot of Total Bill")
plt.show()
```

```python
# 2. Pairplot with Hue Parameter
sns.pairplot(df, hue='sex')
plt.show()
```

```python
# 3. Scatter Plot of Tip vs Total Bill Colored according to Sex
plt.figure(figsize=(8,5))
sns.scatterplot(x='total_bill',
                y='tip',
                hue='sex',
                data=df)
plt.title("Tip vs Total Bill")
plt.show()
```

```python
# 4. Histogram of Total Bill (50 Bins)
plt.figure(figsize=(8,5))
plt.hist(df['total_bill'], bins=50)
plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# 5. Heatmap of Correlation Matrix
plt.figure(figsize=(8,6))
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix,
            annot=True,
            cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
