```python
import numpy as np
```

```python
#1. Write code to create a 4x3 matrix with values ranging from 2 to 13.
arr1 = np.arange(2, 14).reshape(4, 3)
print("1. 4x3 Matrix:")
print(arr1)
```

```python
# 2. Replace odd numbers by -1 in an array created using random.randint(1, 50, 20)
arr2 = np.random.randint(1, 50, 20)
print("Original Array:")
print(arr2)

arr2[arr2 % 2 != 0] = -1
print("Array after replacing odd numbers with -1:")
print(arr2)
```

```python
# 3. Mobile phone prices operations
prices = np.array([6999, 7500, 11999, 27899, 14999, 9999])

# a. Create 1D array
print("Mobile Prices Array:")
print(prices)
```

```python
# b. Convert to float type
prices = prices.astype(float)
print("Converted to Float:")
print(prices)
```

```python
# c. Append new mobile price 13999
prices = np.append(prices, 13999)
print("After Appending 13999:")
print(prices)
```

```python
# d. Reverse the array
prices = prices[::-1]
print("Reversed Array:")
print(prices)
```

```python
# e. Apply GST of 18%
prices = prices * 1.18
print("After Applying 18% GST:")
print(prices)
```

```python
# f. Sort in descending order
prices = np.sort(prices)[::-1]
print("Sorted in Descending Order:")
print(prices)
```

```python
# g. Average mobile phone price
avg_price = np.mean(prices)
print("Average Price:", avg_price)
```

```python
# h. Difference between maximum and minimum price
difference = np.max(prices) - np.min(prices)
print("Difference between Max and Min Price:", difference)
```

```python
# 4. Find indices of non-zero elements
arr4 = np.array([1, 2, 0, 0, 4, 0])
indices = np.nonzero(arr4)
print("\n4. Indices of Non-Zero Elements:")
print(indices)
```
