Q1 Googleplaystore

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/googleplaystore.csv")
```

```python
#a. Display First 5 records from Dataframe
df.head(5)
```

```python
# b. Display category with highest number of apps registered
highest_category = df['Category'].value_counts()
print(highest_category.head(1))
```

```python
# c. What is the average rating for all the apps
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
avg_rating = df['Rating'].mean()
print(avg_rating)
```

```python
# d. Display total number of apps which support Android version 2.3 and up
android_23_apps = df[df['Android Ver'] == '2.3 and up']
print(len(android_23_apps))
```

```python
# e. Impute missing values in Rating column
rating_mean = df['Rating'].mean()
df['Rating'].fillna(rating_mean, inplace=True)
print("\nMissing Values after Imputation:")
print(df['Rating'].isnull().sum())
print("\nMissing Ratings Replaced with Mean =", rating_mean)
```

Q2 Salaries dataset

```python
df = pd.read_csv("/content/Salaries.csv")
```

```python
# a. Check head of dataframe
print(df.head())
```

```python
# b. Find number of entries and columns
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# c. Average Base Pay
df['BasePay'] = pd.to_numeric(df['BasePay'], errors='coerce')
avg_basepay = df['BasePay'].mean()
print(avg_basepay)
```

```python
# d. Highest Overtime Pay
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')
highest_overtime = df['OvertimePay'].max()
print(highest_overtime)
```

```python
# e. Job title of JOSEPH DRISCOLL
job_title = df[df['EmployeeName'] == 'JOSEPH DRISCOLL']['JobTitle']
print(job_title.values)
```

```python
# f. How much does JOSEPH DRISCOLL make(including benefits)?
total_pay = df[df['EmployeeName'] == 'JOSEPH DRISCOLL']['TotalPayBenefits']
print(total_pay.values)
```

```python
# g. Number of unique job titles
unique_jobs = df['JobTitle'].nunique()
print(unique_jobs)
```

```python
# h. Top 5 most common jobs
top_jobs = df['JobTitle'].value_counts().head(5)
print(top_jobs)
```
