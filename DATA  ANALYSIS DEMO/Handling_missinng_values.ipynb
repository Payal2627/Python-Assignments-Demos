```python
import pandas as pd
import numpy as np
```

```python
data = {"eid":[101,102,103,104],"ename":["Kiran","Khushbu","Suraj","Pradeep"],
        "basic":[23000,34000,45000,45000]}
df = pd.DataFrame(data)
df
```

```python
df_new = pd.DataFrame(['Pune','Mumbai','Pune','Mumbai'])
df['location'] = df_new
df
```

```python
df_designation = pd.DataFrame(['Manager','Developer','Developer','Tester'],
                              index = np.arange(105,109))
df_designation
```

```python
df["Designation"] = df_designation
df
```

```python
df_final = pd.concat([df,df_designation],axis=1)
df_final.isna()
```

```python
#Find out total number of missing values in a specific column
df_final.isna().sum()
```

```python
#Delete the missing values, row wise
df_final.dropna()
```

```python
df_final.dropna(axis=1)
df_final
```

```python
df_final['Designation']  = df_final[0]
df_final
```

```python
df_final.drop(0,axis=1)
```

```python
df_final['Designation'].mode() #Output here is a series object
#But we want string
x = df_final['Designation'].mode()[0]
```

```python
df_final['Designation'].fillna(x) #Whenever you will see output it means original data is not changed, new object is created
```

```python
df_final['Designation'].fillna(x,inplace=True)
```
