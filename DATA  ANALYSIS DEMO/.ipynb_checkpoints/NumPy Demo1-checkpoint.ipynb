```python
list1 = [1,2,3,4,5]
list2 = [10,20,30,40,50]
result = list1+list2  
result
```

```python
distance = [100,230,120,340,500]
time = [4,5,6,10,4]
speed = []
for i in range(len(distance)):
  speed.append(distance[i]/time[i])
print(speed)
```

```python
np_distance = np.array(distance)
np_time = np.array(time)
np_speed = np_distance / np_time
np_speed
```
