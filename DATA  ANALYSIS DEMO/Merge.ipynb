```python
import numpy as np
import pandas as pd
```

```python
dept = pd.DataFrame({"deptno":["d101","d102","d103","d104"],
                    "dname":["Accounts","Finance","Marketing","Sales"]})
dept
```

```python
data = {"eid":[101,102,103,104,105,106,107,108],
        "ename":["Kiran","Khushbu","Suraj","Pradeep","Ritu","Rahul","Samay","Tanuja"],
        "basic":[23000,34000,45000,45000,36000,51000,56000,78000],
        "deptno":["d101","d101","d102","d102","d102","d103","d103",np.nan]}
df_emp = pd.DataFrame(data)
df_emp
```

```python
pd.merge(df_emp,dept) #Equjoin or Natural Join
```

```python
data = {"eid":[101,102,103,104,105,106,107,108],
        "ename":["Kiran","Khushbu","Suraj","Pradeep","Ritu","Rahul","Samay","Tanuja"],
        "basic":[23000,34000,45000,45000,36000,51000,56000,78000],
        "d_no":["d101","d101","d102","d102","d102","d103","d103",np.nan]}
df_emp = pd.DataFrame(data)
df_emp
```

```python
#equijoin
pd.merge(df_emp,dept,left_on="d_no",right_on="deptno") #Use this when column names are not name
```

```python
#left join
print(df_emp)
print(dept)
```

```python
#left outer join
pd.merge(df_emp, dept, how="left", left_on="d_no", right_on="deptno")
```

```python
#right outer join
pd.merge(df_emp, dept, how="right", left_on="d_no", right_on="deptno")
```

```python
#full outer join
pd.merge(df_emp, dept, how="outer", left_on="d_no", right_on="deptno")
```
